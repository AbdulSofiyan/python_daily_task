```python
# Create a custom iterator that prints only even numbers from 2 to 20.
numbers = range(2,21,2)
iterator = iter(numbers)
count = 0
for i in range(21):   
    if i % 2 == 0:
       try:
           print(next(iterator))
           count += 1
       except StopIteration:
           break
```
