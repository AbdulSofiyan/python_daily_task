# multiple inheritance

```python
from abc import ABC
```

```python
#  Create two parent classes Father and Mother. Create a child class Child that inherits from both
# Parent 1
class Agent1(ABC):
    def __init__(self, agent_name, agent_role,*args , **kwargs):
        super().__init__(*args, **kwargs)
        self.agent_name = agent_name
        self.agent_role = agent_role

    def get_agent_name(self):
        return self.agent_name

    def get_agent_role(self):
        return self.agent_role


# Parent 2
class Agent2(ABC):
    def __init__(self, agent2_name, agent2_role, *args, **kwargs):
        super().__init__(*args,**kwargs)
        self.agent2_name = agent2_name
        self.agent2_role = agent2_role

    def get_agent2_name(self):
        return self.agent2_name

    def get_agent2_role(self):
        return self.agent2_role


# Child
class Mcp(Agent1, Agent2):
    def __init__(self, agent_name, agent_role,
                 agent2_name, agent2_role, api_key):

        super().__init__(agent_name,agent_role,agent2_name,agent2_role)
        self.api_key = api_key

    def display(self):
        print("Agent 1 Name :", self.agent_name)
        print("Agent 1 Role :", self.agent_role)
        print("Agent 2 Name :", self.agent2_name)
        print("Agent 2 Role :", self.agent2_role)
        print("API Key :", self.api_key)


mcp = Mcp(
    "Flight_agent",
    "Book Flight",
    "Hotel_agent",
    "Book Hotel",
    122344
)

mcp.display()
```

```python
# Define class Sports and class Academic. Create class Student that inherits from both
class Sports:
    def __init__(self, cricket, volleyball, **kwargs):
        super().__init__(**kwargs)
        self.cricket = cricket
        self.volleyball = volleyball


class Academic:
    def __init__(self, year, month, **kwargs):
        super().__init__(**kwargs)
        self.year = year
        self.month = month


class Student(Sports, Academic):
    def __init__(self, cricket, volleyball, year, month, name, age):
        super().__init__(
            cricket=cricket,
            volleyball=volleyball,
            year=year,
            month=month
        )
        self.name = name
        self.age = age

    def display(self):
        print("Sport 1 :", self.cricket)
        print("Sport 2 :", self.volleyball)
        print("Year    :", self.year)
        print("Month   :", self.month)
        print("Name    :", self.name)
        print("Age     :", self.age)


student1 = Student("Cricket", "Volleyball", 2016, "July", "Sofiyan", 21)
student1.display()
```

```python
# two parent and one child 

# parent 1:
class Person:
    
    # constructor
    def __init__(self,name,age,**kwargs): # Person.__init__("sumit",12,**kwargs)
        super().__init__(**kwargs)
        self.name = name
        self.__age = age  # private
        
        
    # getter 
    def get_name(self):
        return f"Person Name : {self.name}"
    
    # getter 
    def get_age(self):
        return f"Person Age : {self.__age}"
    
    # staticmethod
    @staticmethod
    def valid_age(age):
        return 20< age < 50
        
    
    
    # set the age :setter
    def set_age(self,age):
        if Person.valid_age(age):
            self.__age = age
            return f"Person age udpated : {self.__age}"
        
        else :
            print("Please enter the age between the 20 and 50")
            
            
# parent 2 :

class Company:
    
    # constructor:
    def __init__(self,salary,company,**kwargs): # Company.__init__(200,"AM",**kwargs)
        super().__init__(**kwargs)
        self.__salary = salary  # private
        self.company = company
        
        
    # getter 
    def get_salary(self):
        return f"Salary : {self.__salary}"
    
    # static method
    @staticmethod
    def valid_salary(sal):
        return sal <= (20/100)* self.__salary
    
    #setter
    def set_salary(self,sal):
        if Company.valid_salary(sal):
            self.__salary += sal
            return f"Salary is updated : {sal}"
        
        else:
            print("Please enter the salary less than equal to 20% current salary")
            
            
            
# child 
class Employee(Person,Company):
    
    # constructor
    def __init__(self,name,age,salary,company,dept,exp): # Employee.__init__(emp,"sumit",12,2000,"NareshIT","Data Science",6)
        super().__init__(
        name = name,
        age = age,
        salary = salary,
        company = company,
        )
        self.Departemet = dept
        self.__Emp_exp = exp   # private
        
        
        
    # getter 
    def get_exp(self):
        return f"Employee Exp : {self.__Emp_exp}"
    
    # static method 
    @staticmethod
    def valid_exp(exp):
        return 0 < exp < 40
    
    # setter
    def set_exp(self,exp):
        if Employee.valid_exp(exp):
            self.__Emp_exp = exp
            return f"Exployee Exp is Updated  : {self.__Emp_exp}"
        
        else:
            print("Plase enter the exp between the 0 and 40")
            
            
    # all info 
    def All_info(self):
        print(f"Person Name : {self.name}")
        print(f"Person Age : {self.get_age()}")
        print(f"Salary  :{self.get_salary()}")
        print(f"Company Name :{self.company}")
        print(f"Dept Name :{self.Departemet}")
        print(f"Employee Exp :{self.__Emp_exp}")
    
            
# create the class object
emp = Employee("sumit",12,2000,"NareshIT","Data Science",6)

emp.All_info()
```
