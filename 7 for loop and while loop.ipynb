```python
# Count and print how many numbers between 1 to 50 are divisible by 3 using for loop. 
count = 0
for i in range (1,50):
    if i %3 == 0 :
        print(i)
        count+=1  
print(f"count=", count)
        
```

```python
# Write a while loop to print numbers from 1 to 15
num = 1
while num <= 15:
    print(num)
    num+=1
```

```python
# Take a number from user and print its multiplication table using while loop
num = int(input("enter a number :"))
i = 1
while i <= 10 :
    print(num, "x", i, "=", num*i)
    i += 1
    
```

```python
# Use while loop to find the sum of numbers until user enters 0
sum = 0
num = int(input("enter a number :"))
while num != 0:
    sum += num
print(sum)
```

```python
# Print all odd numbers from 1 to 20 using while loop
num = 1
result= []
while num <= 21 :
    if num % 2 != 0:
        result.append("odd")
    else :
        result.append("even")
    num += 1
print(result)
        
        
```
