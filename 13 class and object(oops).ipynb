```python
# Create a class Student with instance attributes name and age. Create two objects and print their details.
class Student:
    def __init__(self,name,age):
        self.Name = name
        self.Age = age
    def display_information(self):
        print(f"Name : {self.Name}")
        print(f"Age : {self.Age}")
std1 = Student("abdul sofiyan",21)
std1.display_information()

    
```

```python
# Define class Car with attributes brand, model, and year. Create an object and display all attributes.
class Car:
    def __init__(self,brand,model,year):
        self.Brand = brand
        self.Model = model
        self.Year = year
    def display_information(self):
        print(f"brand : {self.Brand}")
        print(f"model : {self.Model}")
        print(f"year : {self.Year}")
car1 = Car("verna",16,2026)
car1.display_information()
```

```python
# Create class Book with attributes title, author, price. Add a method display_info().
class Book:
    def __init__(self,title,author,price):
        self.Title = title
        self.Author = author
        self.Price = price
    def display_info(self):
        print(f" title :{self.Title}")
        print(f" author:{self.Author}")
        print(f" price:{self.Price}")
book1 = Book("abcd","efgh",1000)
book1.display_info()
```

```python
# Create class Employee with attributes emp_id, name, salary. Create 3 objects
class Employee:
    def employee_info(self,emp_id,name,salary):
        self.Emp_id  = emp_id
        self.Name = name
        self.salary = salary

    def display_information(self):
        print(f" Emp_id :{self.Emp_id}")
        print(f" Name :{self.Name}")
        print(f" Salary :{self.salary}")
emp1 = Employee()
emp1.employee_info(501,"sofiyan",5000)
emp1.display_information()
print("*" *8)
emp2 = Employee()
emp2.employee_info(502,"abcd",10000)
emp2.display_information()
print("*" *8)
emp3 = Employee()
emp3.employee_info(503,"efgh",15000)
emp3.display_information()
```

```python
# Write a class Mobile with attributes brand, price, color. Print object using __str__ method
class Mobile:
    def __init__(self,brand,price,color):
        self.Brand = brand
        self.Price = price
        self.Color = color
    def __str__(self):
        return f"Brand:{self.Brand}\n, Price:{self.Price}\n, color:{self.Color}
    def display_information(self):
        print(f"brand : {self.Brand}")
        print(f"price : {self.Price}")
        print(f"color : {self.Color}")
mob1 = Mobile("iphone",100000,"white")
mob1.display_information()



```

```python
# Create class Rectangle with attributes length and width. Add method to calculate area.
class Rectangle :
    def __init__(self,length,width):
        self.Length = length
        self.Width = width
        self.calculate_area()
    def calculate_area(self):
        self.area = self.Length*self.Width
    def display_information(self):
        print(f"Length : {self.Length}")
        print(f"Width : {self.Width}")
        print(f"area :{self.area}")
rec1 = Rectangle(10,20)
rec1.display_information()
```

```python
# Define class Person with attributes name and city. Create object and change city value.
class Person:
    def __init__(self,name,city):
        self.Name = name
        self.City = city
    def display_information(self):
        print(self.Name)
        print(self.City)
    def change_City(self,new_city):
        self.City = new_city
        print(f"city name {self.City} updated successfully")
obj1 = Person("sofiyan","hyd")
obj1.display_information()
obj1.change_City("banglore")
obj1.display_information()

    
```
