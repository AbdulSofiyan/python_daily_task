```python
# Define class Shape with method area(). Create subclasses Circle and Rectangle
class Shape:
    def area(self):
        pass
class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        print(f"Circle area : {3.14*(self.radius**2)}")
class Rectangle(Shape):
    def __init__(self,length,width):
        self.length = length
        self.width = width
    def area(self):
        print(f"Rectangle area : {self.length*self.width}")

circle1 = Circle(10)
circle1.area()
rectangle1 = Rectangle(2,1)
rectangle1.area()

        
```

```python
# Create class Employee with attributes emp_id and salary. Create subclasses Manager and Developer.
class Employee:
    def __init__(self,emp_id,salary):
        self.emp_id = emp_id
        self.salary = salary
class Manager(Employee):
    def Manager_details(self):
        print(f"manager id :{self.emp_id}")
        print(f"manager salary :{self.salary}")
class Developer(Employee):
    def Developer_details(self):
        print(f"developer id :{self.emp_id}")
        print(f"developer salary :{self.salary}")
man = Manager("sofiyan",200000)
deve = Developer("sumith",250000)

man.Manager_details()
deve.Developer_details()
       

```

```python
# BankAccount → SavingsAccount, CurrentAccount . Define class BankAccount with method deposit().
class Bank_account:
    def deposit(self):
        pass
#child class:1
class Saving_account(Bank_account):
    def __init__(self,bal):
        self.bal = bal
    def deposit_details(self,amt):
        if amt > 0 :
            self.bal = self.bal+amt
            print(f"savings balance :",self.bal)
        else :
            print("enter correct value")
    
            
#child class:2
class CurrentAccount(Bank_account):
    def __init__(self,bal1):
        self.bal1 = bal1
    def deposit_details1(self,amt):
        if amt > 0 :
            self.bal1 = self.bal1+amt
            print(f"current balance :",self.bal1)
sav = Saving_account(10000)
sav.deposit_details(1000)
curr = CurrentAccount(20000)
curr.deposit_details1(2000)
        
```

```python
class Employee:
    total_emp = 0

    def __init__(self, name, salary, dept):
        self.__name = name
        self.__salary = salary
        self.__dept = dept
        Employee.total_emp += 1

    # Getters
    def get_name(self):
        return self.__name

    def get_salary(self):
        return self.__salary

    def get_dept(self):
        return self.__dept

    # Static Method
    @staticmethod
    def valid_salary(salary):
        return salary > 0

    # Setter
    def set_salary(self, salary):
        if Employee.valid_salary(salary):
            self.__salary = salary
            print("Salary Updated Successfully")
        else:
            print("Invalid Salary")

    # Class Method
    @classmethod
    def get_total(cls):
        return cls.total_emp

    # Instance Method
    def work(self):
        print("Employee is working")

    def calculate_bonus(self):
        return self.__salary * 0.05

    def show(self):
        print("Name :", self.get_name())
        print("Salary :", self.get_salary())
        print("Department :", self.get_dept())
        print("Bonus :", self.calculate_bonus())


# Child Class 1
class Developer(Employee):
    def __init__(self, name, salary, dept, language):
        super().__init__(name, salary, dept)
        self.language = language

    def work(self):
        print("Developer is writing code")

    def calculate_bonus(self):
        return self.get_salary() * 0.20

    def show(self):
        super().show()
        print("Programming Language :", self.language)


# Child Class 2
class Designer(Employee):
    def __init__(self, name, salary, dept, tool):
        super().__init__(name, salary, dept)
        self.tool = tool

    def work(self):
        print("Designer is creating UI")

    def calculate_bonus(self):
        return self.get_salary() * 0.15

    def show(self):
        super().show()
        print("Design Tool :", self.tool)


# Child Class 3
class Manager(Employee):
    def __init__(self, name, salary, dept, team_size):
        super().__init__(name, salary, dept)
        self.team_size = team_size

    def work(self):
        print("Manager is managing the team")

    def calculate_bonus(self):
        return self.get_salary() * 0.15

    def show(self):
        super().show()
        print("Team Size :", self.team_size)


# Objects
d = Developer("Sofiyan", 50000, "IT", "Python")
de = Designer("virat", 45000, "abcd", "java")
m = Manager("kohli", 70000, "HR", 10)

print("------ Developer ------")
d.show()
d.work()

print("\n------ Designer ------")
de.show()
de.work()

print("\n------ Manager ------")
m.show()
m.work()

print("\nTotal Employees :", Employee.get_total())
```
