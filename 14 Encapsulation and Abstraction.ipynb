```python
# Create a class named BankAccount.
class Bank_account:
    def __init__(self, owner, bal):
        self.__Owner = owner      # Private attribute
        self.__Bal = bal          # Private attribute

    # Getter methods
    def deposit_amt(self,amt):
        self.__Bal += amt

    def withdraw_amt(self,amt):
        self.__Bal -= amt
        

    # Setter method
    def set_deposit_amt(self, amt):
        if amt > 0:
            self.__Bal += amt
        else:
            print("Invalid")
    def set_withdraw_amt(self, amt):
        if amt > 0 and amt <= self.__Bal :
            self.__Bal -= amt
        else:
            print("Invalid")

    # Instance method
    def display_bank_details(self):
        print(f"Owner : {self.__Owner}")
        print(f"Balance : {self.__Bal}")


# Create object
bank1 = Bank_account("Sofiyan", 10000)

# Display details
bank1.display_bank_details()

# Update balance using setter
bank1.set_deposit_amt(15000)

print("After Updating Balance:")
bank1.display_bank_details()

bank1.set_withdraw_amt(1000)
print("After Updating Balance:")
bank1.display_bank_details()
```

```python
# Create a class named Student.
class Student:
    def __init__(self,name,marks):
        self.__Name = name
        self.__Marks = marks
    def get_Name(self):
        return self.__Name
    def get_Marks(self):
        return self.__Marks
    def set_marks(self,marks):
        if marks >= 0 and marks <= 100:
            self.__Marks = marks
        else:
            print("invalid marks")
    def display_all_details(self):
        print(f" Name : {self.__Name}")
        print(f" Marks : {self.__Marks}")

std1 = Student("sofiyan",50)
std1.display_all_details()
std1.set_marks(90)
std1.display_all_details()
        
```

```python
# Create a class named Employee. Salary cannot decrease by more than 10% of the current salary.
class Employee:
    def __init__(self, name, salary):
        self.__Name = name
        self.__Salary = salary

    def get_salary(self):
        return self.__Salary

    def set_salary(self, new_salary):
        allow = self.__Salary * 0.90

        if new_salary < allow:
            print("Reject")
        else:
            self.__Salary = new_salary

    def display_all_details(self):
        print(f"Name : {self.__Name}")
        print(f"Salary : {self.__Salary}")


emp = Employee("sofiyan", 50000)
emp.display_all_details()

emp.set_salary(5000)
emp.display_all_details()

emp.set_salary(47000)
emp.display_all_details()
        
        


```

```python
# Create a class named Mobile , Price should always be positive.
class Mobile:
    def __init__(self, name, price):
        self.__Name = name
        self.__Price = price

    # Getter
    def get_price(self):
        return self.__Price

    # Setter
    def set_price(self, new_price):
        if new_price > 0:
            self.__Price = new_price
        else:
            print("Reject")

    # Display Method
    def display_all_details(self):
        print(f"Mobile Name : {self.__Name}")
        print(f"Mobile Price: {self.__Price}")


# Create object
mob = Mobile("iPhone", 50000)

# Display details
mob.display_all_details()

# Update price
mob.set_price(-47000)

# Display updated details
mob.display_all_details()
```

```python
# take the parent ; company
# get private data
#make two abstract method:
#1 . work
#2. salary
# display all info : name , role, work, salary

# Parent class
class Company(ABC):

    # Constructor
    def __init__(self, name, salary, role):
        self.__Name = name
        self.__Salary = salary
        self.__Role = role

    # Getter methods
    def get_name(self):
        return self.__Name

    def get_salary(self):
        return self.__Salary

    def get_role(self):
        return self.__Role

    # Abstract methods
    @abstractmethod
    def work(self):
        pass

    @abstractmethod
    def salary(self):
        pass

    # Display all details
    def display_all_details(self):
        print(f"Name   : {self.get_name()}")
        print(f"Role   : {self.get_role()}")
        print("Work   :", end=" ")
        self.work()
        print("Salary :", self.salary())


# Child class 1
class Manager(Company):

    def __init__(self, name, salary):
        super().__init__(name, salary, "Manager")

    def work(self):
        print("Manages the entire team")

    def salary(self):
        return self.get_salary()


# Child class 2
class Employee(Company):

    def __init__(self, name, salary):
        super().__init__(name, salary, "Employee")

    def work(self):
        print("Works for the company")

    def salary(self):
        return self.get_salary()


# Create objects
m1 = Manager("Sofiyan", 100000)
e1 = Employee("Rahul", 30000)

# Display details
m1.display_all_details()
print()
e1.display_all_details()
```
