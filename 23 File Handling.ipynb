```python
# Count the number of lines in a text file.
with open("abcd.txt","r") as file:
    line_count = len(file.readline())
    print(f"number of lines in file:{line_count}")
        
    
```

```python
# Count the number of words in a text file.
with open("abcd.txt","r") as file:
    word_count = len(file.read().split())
    print(f"number of word in file:{word_count}")
```

```python
# Count the number of characters in a text file
with open("abcd.txt","r") as file:
    character_count = len(file.read())
    print(f"number of characters in file:{character_count}")
```

```python
# Read a file line by line using a loop.
file = open("abcd.txt","r")
count = 0
for i in file:
    count += 1
print(count)
```

```python
# Write a program to copy content from one file to another file.
with open("fruits1.txt","w")as file1:
    file1.write("fruit1 : apple \nfruit2 : banana")
with open("fruits1.txt","r")as file1:
    content = file1.read()

with open("fruits2.txt","w")as file2:
    file2.write(content)
with open("fruits2.txt","r") as file2:
    print(file2.read())
    
    
```

```python
# Write a program to copy content from one file to another file.
with open("fruits1.txt","w")as file1:
    file1.write(f"file1 : fruit1 : apple ,fruit2 : banana")
with open("fruits1.txt","r")as file1:
    content = file1.read()
    print(content)

with open("fruits2.txt","w")as file2:
    file2.write(content)
with open("fruits2.txt","r") as file2:
    print(file2.read())
```
