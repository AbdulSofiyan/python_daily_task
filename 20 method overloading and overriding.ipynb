# method overloading

```python
# Create a Calculator class that can add 2 numbers or 3 numbers using the same method name.
class Calculator:
    def add(self,a,b,c=0):
        return a+b+c
a1 = Calculator()
print(a1.add(1,2))
print(a1.add(1,2,3))


```

```python
# Create an Area class that calculates the area of a square and rectangle using the same method.
class Area:
    def Calculate(self,length,width = 0):
        if width == 0 :
            return length*length
        else:
             return length*width
a = Area()
print("square area :",a.Calculate(5))
print("rectangle area:",a.Calculate(5,10))
        
```

```python
# Create a Student class with a details() method that accepts different numbers of arguments.
class Student:
    def details(self,name,age = 0,city = 0,addr = 0):
        return name,age,city,addr

std1 = Student()
print(std1.details("sofiyan",21))
```

```python
import math
# Create a MathOperations class that multiplies any number of values.
class Mathoperation:
    def mul (self,a,b,*args):
        result =  a*b
        for num in args:
            result *= num
        return result
        

a1 = Mathoperation()
print(a1.mul(1,2))
print(a1.mul(1,2,3,4))
```

```python
# Create a Display class that prints different messages based on the argument type.
class Display:
    def show(self,data):
        if isinstance(data,int):
            print("this is integer")
        elif isinstance(data,str):
            print("this is string")
        elif isinstance(data,list):
            print("this is list")
        else :
            print("unknown type")

obj1 = Display()
obj1.show(12.34)
```

```python
# Create a Volume class that calculates the volume of a cube and cuboid
class Volume:
    def Calculate(self,length,width = 0,height = 0):
        if width == 0 and height == 0:
            return length*length*length
        else:
             return length*width*height
a = Volume()
print("cube :",a.Calculate(5))
print("cuboid:",a.Calculate(5,10,15))
```

# method overriding

```python
# child class change the behaviour of parent classs method.
class Animal:
    def __init__(self,name):
        self.name = name
    # method1
    def sound(self):
        print(f"{self.name}make sound:")

    # method2
    def move(self):
        print(f"{self.name} move :")

# overriding the method using child class
class Cat(Animal):
    def sound(self):
        print(f"{self.name} sound:meowww")
    def move(self):
        print(f"{self.name} is running")

class Dog(Animal):
    def sound(self):
        print(f"{self.name} sound:bowwww")
    def move(self):
        print(f"{self.name} is attack")

# class1 object
cat1 = Cat("siri")
dog1 = Dog("goli")

cat1.sound()
cat1.move()
dog1.sound()
dog1.move()



def make_sound(Animal):
    animal.sound()

def make_move(Animal):
    animal.move()

print(cat1.sound())
print(cat1.move())
print(dog1.sound())
print(dog1.move())

animal = [cat1,dog1]
for animal in animal:
    animal.sound()
    animal.move()
```
