```python
# Write a lambda function to return the square of a number.
squared = lambda x: x*x
squared(5)
```

```python
# Write a lambda function to return the cube of a number
cubic = lambda x: pow(x,3)
cubic(4)
```

```python
# Write a lambda function to add two numbers
add = lambda x,y : x + y
add(3,4)
```

```python
# Write a lambda function to find the maximum of two numbers
maxi = lambda x,y : x if x>y else y
print(maxi(15,8))
```

```python
# Write a lambda function that returns "Even" if a number is even; otherwise "Odd".
result = lambda x : "even" if x  % 2==0 else "odd" 
result(9)
```

```python
# Use filter() with lambda to keep only even numbers
numbers = [1,2,3,4,5,6]
even = list(filter( lambda x : x %2 == 0 , numbers))
even
```

```python
# Use filter() with lambda to keep numbers greater than 50.
numbers = [20,30,55,67,32]
a = list(filter( lambda x : x > 50, numbers))
a
```

```python
# Sort a list of numbers in ascending order using lambda
numbers = [9,5,1,8,2]
sort = sorted(numbers, key = lambda x : x)
sort
```

```python
# Sort a list of numbers in descending order using lambda
numbers = [9,5,1,8,2]
sort1 =list( reversed(sorted(numbers, key = lambda x : x)))
sort1
```

```python
from functools import reduce
```

```python
# Use reduce() with lambda to find the sum of all numbers
numbers = [1,2,3,4,5,6,7,8]
sum1 = reduce(lambda x,y : x+y, numbers)
sum1
```

```python
# Use reduce() with lambda to find the product of all numbers

numbers = [1,2,3]
product = reduce(lambda x,y : x*y, numbers)
product
```

```python
# Sort the following list according to age
students = [
("Ram",22),
("Amit",19),
("Neha",25),
("Raj",20)
]
sort = list(sorted(students, key = lambda x : x[1]))
sort
```

```python
# Sort the following dictionary by values
marks = {
"A":80,
"B":95,
"C":70,
"D":88
}
sort1 = dict(sorted(marks.items(), key = lambda x : x[1]))
print(sort1)
```
