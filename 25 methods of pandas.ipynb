# operators and loops

```python
#Take two numbers from user. Perform all arithmetic operations and also check which number is greater using comparison operators.
a= int(input("enter a number"))
b= int(input("enter a number"))
print(a+b)
print(a-b)
print(a%b)
print(a*b)
print(a==b)
print(a!=b)
print(a>=b)
print(a<=b)

```

```python
# Print all numbers from 1 to 50. Skip multiples of 5 using continue and stop completely when number 40 is reached using break.
for i in range(1,51):
    if i == 40:
         break
    if i%5 == 0:
         continue
        
    print(i)
   
```

```python
# Take numbers from user until they enter negative number. Calculate sum and average using while loop.
start = 0
while True:
    a= int(input("enter a number"))
    if a <= 0:
        break
    start = a+start
print(start)
    
```
