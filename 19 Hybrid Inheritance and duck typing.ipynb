```python
# 1. Create a hybrid inheritance structure with classes A → B and C → D (where D inherits from both B and C).
class A:
    def __init__(self,name,age,**kwargs):
        self.name = name
        self.age = age
        super().__init__(**kwargs)
class B(A):
    def __init__(self,weight,**kwargs):
        self.weight = weight
        super().__init__(**kwargs)
class C:
    def __init__(self,height,**kwargs):
        self.height = height
        super().__init__(**kwargs)
class D (B,C):
    def __init__(self,name,age,weight,height,education):
        super().__init__(
            name = name,
            age = age,
            weight = weight,
            height = height
        )
        self.education = education
  
    def show_all(self):
        print("\n----- D Details -----")
        print("Name           :", self.name)
        print("Age            :", self.age)
        print("weight        :", self.weight)
        print("height   :", self.height)
        print("education          :", self.education)

obj1 = D("sofiyan",20,60,6,"graduation")
obj1.show_all()
```

```python
# Create Person, Employee, Student, and TeachingAssistant using hybrid inheritance
class Person:
    def __init__(self,name,age,**kwargs):
        self.name = name
        self.age = age
        super().__init__(**kwargs)
class Employee(Person):
    def __init__(self,salary,**kwargs):
        self.__salary = salary
        super().__init__(**kwargs)
    def get_salary(self):
        return f"employee salary :{self.__salary}"
class Student(Person):
    def __init__(self,marks,**kwargs):
        self.marks = marks
        super().__init__(**kwargs)
class Teaching_assistance (Employee,Student):
    def __init__(self,name,age,salary,marks,dept):
        super().__init__(
            name = name,
            age = age,
            salary = salary,
            marks = marks
        )
        self.dept = dept
  
    def show_all(self):
        print("\n----- Teaching assistance Details -----")
        print("Name           :", self.name)
        print("Age            :", self.age)
        print("weight        :", self.get_salary())
        print("height   :", self.marks)
        print("dept          :", self.dept)

obj1 = Teaching_assistance("sofiyan",20,30000,90,"cse")
obj1.show_all()
```

```python
# Create Vehicle → Car, Electric → ElectricCar (ElectricCar inherits from both).

class Vechile:
    def __init__(self,name,**kwargs):
        self.name = name
        super().__init__(**kwargs)
class Car(Vechile):
    def __init__(self,model,**kwargs):
        self.model = model
        super().__init__(**kwargs)
class Electric:
    def __init__(self,price,**kwargs):
        self.price = price
        super().__init__(**kwargs)
class Electric_car (Car,Electric):
    def __init__(self,name,model,price,capacity):
        super().__init__(
            name = name,
            model = model,
            price = price,
        )
        self.capacity = capacity
  
    def show_all(self):
        print("\n----- D Details -----")
        print("Name           :", self.name)
        print("model            :", self.model)
        print("price        :", self.price)
        print("capacity   :", self.capacity)
        

obj1 = Electric_car("verna",2019,100000,"five")
obj1.show_all()
```

```python
# What is Duck Typing? Explain with a simple example
class Creditcard :
    def pay(self,amount):
        return f"paying amount using credit card : {amount}"
                                
class Debitcard :
    def pay(self,amount):
        return f"paying amount using debit card : {amount}"

class Upi :
    def pay(self,amount):
        return f"paying amount using Upi : {amount}"

# make functions

def make_payments(payment,amount):
    print(payment.pay(amount))

# create the object
cr = Creditcard()
de = Debitcard()
upi = Upi()
# call function
make_payments(cr,500)
make_payments(de,500)
make_payments(upi,500)
```
