```python
from abc import ABC, abstractmethod
```

```python
# Create an abstract class named Shape
class Shape(ABC):
    
    @abstractmethod
    def area(self):
        pass
    def perimeter(self):
        pass

# child class
class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        print("area of circle =",3.14*self.radius**2)
    def perimeter(self):
        print("perimeter of circle =",2*3.14*self.radius)
        
class Rectangle(Shape):
    def __init__(self,length,width):
        self.length = length
        self.width = width
        
    def area(self):
        print("area of rectangle =",self.length*self.width)
    def perimeter(self):
        print("perimeter of rectangle =",2*(self.length + self.width))
#object class attributes
circle1 = Circle(5)
circle1.area()
circle1.perimeter()
print("*****")
rectangle1 = Rectangle(5,6)
rectangle1.area()
rectangle1.perimeter()
    
    
        
```

```python
# Create an abstract class named Vehicle.
class Vechile(ABC):
    def __init__(self):
        pass
    @abstractmethod
    def start(self):
        pass
    def stop(self):
        pass
# child class
class Car(Vechile):
    
    def start(self):
        print("car is started")
    def stop(self):
        print("car is stopped")

car1 = Car()
car1.start()
car1.stop()
```

```python
# Create an abstract class named Payment.
class Payments (ABC):
    @abstractmethod
    def process_payment(self,amount):
        pass
#child class:1
class Credit_card(Payments):
    def process_payment(self,amount):
        print(f"credit card payment is successful")
        print(f"amount paid:{amount}")
#child class:2
class Cash(Payments):
    def process_payment(self,amount):
        print(f"cash payment is successful")
        print(f"amount paid:{amount}")
#object class
cred = Credit_card()
cred.process_payment(5000)
print()
cash = Cash()
cash.process_payment(2500)

    
```

```python
class Company(ABC):

    # Constructor
    def __init__(self, name, salary, role):
        self.__Name = name
        self.__Salary = salary
        self.__Role = role

    # Getter methods
    def get_name(self):
        return self.__Name

    def get_salary(self):
        return self.__Salary

    def get_role(self):
        return self.__Role

    # Abstract methods
    @abstractmethod
    def work(self):
        pass

    @abstractmethod
    def salary(self):
        pass

    # Display all details
    def display_all_details(self):
        print(f"Name   : {self.get_name()}")
        print(f"Role   : {self.get_role()}")
        print("Work   :", end=" ")
        self.work()
        print("Salary :", self.salary())


# Child class 1
class Manager(Company):

    def __init__(self, name, salary):
        super().__init__(name, salary, "Manager")

    def work(self):
        print("Manages the entire team")

    def salary(self):
        return self.get_salary()


# Child class 2
class Employee(Company):

    def __init__(self, name, salary):
        super().__init__(name, salary, "Employee")

    def work(self):
        print("Works for the company")

    def salary(self):
        return self.get_salary()


# Create objects
m1 = Manager("Sofiyan", 100000)
e1 = Employee("fahad", 30000)

# Display details
m1.display_all_details()
print()
e1.display_all_details()
```

```python
from abc import ABC, abstractmethod

class Agent(ABC):
    def __init__(self, agent_name, agent_apikey):
        self.Agent_name = agent_name
        self.__Agent_apikey = agent_apikey

    def get_api_key(self):
        return f"Agent_api_key : {self.__Agent_apikey}"

    @abstractmethod
    def LLM_model(self):
        pass

    def agent_info(self):
        return f"Agent name : {self.Agent_name}\nAgent api key : {self.__Agent_apikey}"


# Child class 1
class Flight_agent(Agent):
    def __init__(self, agent_name, agent_apikey, role):
        super().__init__(agent_name, agent_apikey)
        self.Agent_role = role

    def LLM_model(self):
        return "OpenAI LLM Model"

    def Display_details(self):
        print(f"Agent role : {self.Agent_role}")
        print(self.agent_info())
        print(self.LLM_model())


# Child class 2
class Hotel_agent(Agent):
    def __init__(self, agent_name, agent_apikey, agent_role):
        super().__init__(agent_name, agent_apikey)
        self.Hotel_agent_role = agent_role

    def LLM_model(self):
        return "Grok AI"

    def Display_details(self):
        print(f"Agent role : {self.Hotel_agent_role}")
        print(self.agent_info())
        print(self.LLM_model())


# Create objects
flight = Flight_agent("abcdef", 123456, "Book the flight")
hotel = Hotel_agent("adsdsd", 56789, "Book the hotel")

# Access details
print(flight.get_api_key())
print(flight.agent_info())
flight.Display_details()

print("\n-----------------\n")

hotel.Display_details()
```
