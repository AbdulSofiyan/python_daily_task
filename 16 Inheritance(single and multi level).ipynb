```python
from abc import ABC, abstractmethod
```

```python
class Animals(ABC):
    #abstract method:
    @abstractmethod
    def sound(self):
        pass
#child class:1
class Cat(Animals):
    def sound(self):
        print("meoww")
#child class:2
class Dog(Animals):
    def sound(self):
        print("bhauu")
#child class:3
class Cow(Animals):
    def sound(self):
        print("mooo") 
# create the class object
cat = Cat()
dog = Dog()
cow = Cow()

obj = Animals
cat.sound()
dog.sound()
cow.sound()
```

```python
class Person:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    def get_name(self):
        return self.__name

    def get_age(self):
        return self.__age

    def set_age(self, age):
        if 20 <= age <= 60:
            self.__age = age
            print("Age updated.")
        else:
            print("Age must be between 20 and 60.")

    def person_info(self):
        return f"Name : {self.__name}\nAge : {self.__age}"


class Employee(Person):

    def __init__(self, name, age, com_name, emp_salary):
        super().__init__(name, age)
        self.__com_name = com_name
        self.__emp_salary = emp_salary

    def get_company(self):
        return self.__com_name

    def get_salary(self):
        return self.__emp_salary

    def set_emp_salary(self, emp_salary):
        self.__emp_salary = emp_salary
        print("Salary updated.")

    def employee_info(self):
        return (
            f"{self.person_info()}\n"
            f"Company : {self.__com_name}\n"
            f"Salary : {self.__emp_salary}"
        )


class Manager(Employee):

    def __init__(self, name, age, com_name, emp_salary, team_size, bonus):
        super().__init__(name, age, com_name, emp_salary)
        self.__team_size = team_size
        self.__bonus = bonus

    def get_team_size(self):
        return self.__team_size

    def get_bonus(self):
        return self.__bonus

    def set_team_size(self, team_size):
        self.__team_size = team_size

    def set_bonus(self, bonus):
        self.__bonus = bonus

    def manager_info(self):
        return (
            f"{self.employee_info()}\n"
            f"Team Size : {self.__team_size}\n"
            f"Bonus : {self.__bonus}"
        )


# Driver Code
m1 = Manager("Sofiyan", 21, "TCS", 80000, 10, 15000)

print(m1.manager_info())

print()

print("Name:", m1.get_name())
print("Age:", m1.get_age())
print("Company:", m1.get_company())
print("Salary:", m1.get_salary())
print("Team Size:", m1.get_team_size())
print("Bonus:", m1.get_bonus())

print()

m1.set_age(40)
m1.set_emp_salary(90000)
m1.set_bonus(20000)

print()
print(m1.manager_info())
```
