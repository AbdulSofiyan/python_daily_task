```python
# Remove element 20 from set s = {10, 20, 30, 40} using remove() and
#discard(). Show difference.

s = {10,20,30,40}
s.remove(20)
s.discard(20)
s
# in remove() if the value which we have to remove is not found, then it will show error. 
# in discard () if the value which we have to remove is not found it will not return anything(none)
```

```python
#Create two sets A = {1, 2, 3, 4} and B = {3, 4, 5, 6}. Print union, 
#intersection, and difference. 

A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print(A.union(B))
print(A.intersection(B))
print(A.difference(B))
```
